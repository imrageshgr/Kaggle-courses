# Functions

```python
# Function define by def keyword
# Syntax:
def function_name(parameters):
    # function body
    pass
```

```python
# Example of a function
def add(a, b):
    return a + b
```

```python
# Function calling
# Syntax:
result = function_name(...)
```

```python
# Example of a function calling
result = add(2, 3)
```

```python
print(result)
```

```python
# Function without parameters
def greet():
    print("Hello, World!")
    return "Greeting sent!"
```

```python
greet()  # Calling the function without parameters
```

```python
print(greet())  # Calling the function and printing the return value
```

```python
# Function with multiple parameters
def multiply(a, b, c, d):
    return a * b * c * d

# Example of a function calling
result = multiply(4, 5, 6, 7)
print(result)
```
