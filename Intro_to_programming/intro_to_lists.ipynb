# Intro to Lists

```python
my_collection = [1, 3,5, 7, 9] # list define by this [] symbol.
```

```python
type(my_collection)
```

```python
len(my_collection) # Check list length.
```

```python
my_collection[0] # Value access by index at staring order.
```

```python
my_collection[-1] # access by reverse order of list.
```

```python
# slicing

my_collection[:2] # Print starting first two values.
```

```python
my_collection[-3:] # End of last three values.
```

```python
my_collection
```

```python
# add value to list.
my_collection.append(11)
my_collection.append(111)
my_collection
```

```python
# removing value from list.
my_collection.remove(111)
```

```python
my_collection
```

```python
len(my_collection) # Length of list.
```

```python
min(my_collection) # minimum value of list.
```

```python
max(my_collection) # Maximum value of list.
```

```python
sum(my_collection) # Sum of list.
```
