# Conditions and Conditional Statements

## Condition Statements

```python
# Conditions is either True or False
condition = 4 > 1
print(condition) 
```

```python
condition = 4 < 1
print(condition)
```

```python
a = 8
b = 5
```

```python
c = a > b # Greater than
print(c)
```

```python
c = a < b # Less than
print(c)
```

```python
c = a == b # Equal to
print(c)
```

```python
c = a >= b # Greater than or equal to
print(c)
```

```python
c = a <= b # Less than or equal to
print(c)
```

```python
c = a !=  b # Not equal to
print(c)
```

# Conditional Statements

```python
# if statement
# if statement is use for single conditional statements. With combination of condition statements.
if a > b:
    print("It's correct") # when the condition is true, this statement will be executed.
print("A is greater than B: ", a, b)
```

```python
if a < b:
    print("It's correct") # when the condition is false, this statement will not be executed.
print("A is less than B.:", a, b)
```

```python
def check_number(n):
    if n > 0: # if statement in function
        print("It's a positive number") 
```

```python
check_number(5)
```

```python
check_number(-1)
```

```python
# if else statement
# if else statement is use for two conditional statements. With combination of condition statements.
if a > b:
    print("It's correct") # when the condition is true, this statement will be executed.
else:   
    print("It's not correct") # when the condition is false, this statement will be executed.

print("A is greater than B.: ", a, b)
```

```python
if a > b:
    print("It's correct") # when the condition is true, this statement will be executed.
else:   
    print("It's not correct") # when the condition is false, this statement will be executed.

print("A is greater than B.: ", a, b)
```

```python
# elif statement
# elif statement is use for multiple conditional statements. With combination of condition statements.
if a > b:
    print("It's correct") # when the condition is true, this statement will be executed.
elif a < b:
    print("It's not correct") # when the condition is false, this statement will be executed.
else:
    print("It's equal") # when the condition is false, this statement will be executed.
```
