# Data Types

```python
# Data type define the which type of data we are going to use in our program. 
# And variable store the data that define the which data is store in that variable.

height = 5.8
```

```python
print(height)
```

```python
# Check the data type of the variable height
# By using the type() function.
print(type(height))
```

## Integers

```python
age = 21
type(age)
```

```python
print(type(age))
```

## Floats

```python
weight = 70.5
type(weight)
```

```python
print(type(weight))
```

```python
pi = 22/7
```

```python
print(pi)
```

```python
# round() function is used to round the number to the nearest integer.
print(round(pi))
```

```python
# Also fix the number of decimal places by using round() function.
print(round(pi, 2))
print(round(pi, 4))
```

## Booleans

```python
# It have only two values True and False.
is_student = True
is_employee = False

type(is_student)
type(is_employee)
```

```python
print(type(is_student))
print(type(is_employee))
```

```python
# not operator is used to reverse the value of the boolean variable.
is_student = not True
print(type(is_student))
```
